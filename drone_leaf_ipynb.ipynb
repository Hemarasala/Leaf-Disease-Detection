```python
++# ================================================================
#   🛿 UAV PLANT DISEASE DETECTION SYSTEM
#   Complete Google Colab Code — Copy & Run Cell by Cell
#   Dataset: PlantVillage (via TensorFlow Datasets — no download!)
#   Model: MobileNetV2 Transfer Learning + Fine-tuning
#   Output: Trained model + TFLite for Raspberry Pi
# ================================================================


# ───────────────────────────────────
# CELL 1 — INSTALL DEPENDENCIES
# ───────────────────────────────────
!pip install tensorflow-datasets tensorflow opencv-python-headless matplotlib seaborn pillow -q


# ───────────────────────────────────
# CELL 2 — IMPORTS
# ───────────────────────────────────
import os
import json
import random
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns
import cv2
import warnings
warnings.filterwarnings("ignore")

import tensorflow as tf
import tensorflow_datasets as tfds
from tensorflow.keras import layers, models, optimizers, callbacks
from tensorflow.keras.applications import MobileNetV2
from sklearn.metrics import classification_report, confusion_matrix
print("Hello")
print(f"✅ TensorFlow  : {tf.__version__}")
# Safely check for tfds.__version__
if hasattr(tfds, '__version__'):
    print(f"✅ TFDS        : {tfds.__version__}")
else:
    print("✅ TFDS        : Version information not available, but module loaded.")
print(f"✅ GPU Available: {len(tf.config.list_physical_devices('GPU')) > 0}")


# ───────────────────────────────────
# CELL 3 — CONFIG
# ───────────────────────────────────
CONFIG = {
    "IMG_SIZE"         : 224,
    "BATCH_SIZE"       : 32,
    "TRAIN_SAMPLES"    : 4000,   # ~5000 total
    "VALID_SAMPLES"    : 1000,
    "EPOCHS_FROZEN"    : 10,     # Phase 1: train head only
    "EPOCHS_FINETUNE"  : 15,     # Phase 2: fine-tune deeper layers
    "LEARNING_RATE"    : 1e-3,
    "FINETUNE_LR"      : 1e-5,
    "MODEL_PATH"       : "plant_disease_model.h5",
    "TFLITE_PATH"      : "plant_disease_model.tflite",
    "CLASS_JSON"       : "class_indices.json",
    "SEED"             : 42,
}

tf.random.set_seed(CONFIG["SEED"])
np.random.seed(CONFIG["SEED"])


# ───────────────────────────────────
# CELL 4 — COMPLETE DISEASE INFO DATABASE (All 38 Classes)
# ───────────────────────────────────
DISEASE_DB = {
    # ── APPLE (4) ──
    "Apple___Apple_scab"                                    : ("Apple",      "Apple Scab",                "Moderate",  "Fungal (Venturia inaequalis)",          "Apply Captan/Mancozeb fungicide. Remove infected leaves.",              0.65),
    "Apple___Black_rot"                                     : ("Apple",      "Black Rot",                 "Severe",    "Fungal (Botryosphaeria obtusa)",         "Prune infected branches. Use copper-based fungicide.",                  0.85),
    "Apple___Cedar_apple_rust"                              : ("Apple",      "Cedar Apple Rust",          "Moderate",  "Fungal (Gymnosporangium)",               "Apply myclobutanil during wet spring season.",                         0.50),
    "Apple___healthy"                                       : ("Apple",      "Healthy",                   "Healthy ✅","None",                                  "No treatment needed. Maintain good irrigation.",                       0.00),

    # ── BLUEBERRY (1) ──
    "Blueberry___healthy"                                   : ("Blueberry",  "Healthy",                   "Healthy ✅","None",                                  "No treatment needed.",                                                  0.00),

    # ── CHERRY (2) ──
    "Cherry_(including_sour)___Powdery_mildew"              : ("Cherry",     "Powdery Mildew",            "Moderate",  "Fungal (Podosphaera clandestina)",       "Apply sulfur-based fungicide. Improve air circulation.",                0.55),
    "Cherry_(including_sour)___healthy"                     : ("Cherry",     "Healthy",                   "Healthy ✅","None",                                  "No treatment needed.",                                                  0.00),

    # ── CORN (4) ──
    "Corn_(maize)___Cercospora_leaf_spot Gray_leaf_spot"    : ("Corn",       "Gray Leaf Spot",            "Moderate",  "Fungal (Cercospora zeae-maydis)",        "Apply strobilurin fungicides. Use resistant hybrids.",                  0.60),
    "Corn_(maize)___Common_rust_"                           : ("Corn",       "Common Rust",               "Moderate",  "Fungal (Puccinia sorghi)",               "Apply triazole fungicides. Use resistant varieties.",                   0.60),
    "Corn_(maize)___Northern_Leaf_Blight"                   : ("Corn",       "Northern Leaf Blight",      "Moderate",  "Fungal (Exserohilum turcicum)",          "Apply fungicide at silking. Use tolerant varieties.",                   0.65),
    "Corn_(maize)___healthy"                                : ("Corn",       "Healthy",                   "Healthy ✅","None",                                  "No treatment needed.",                                                  0.00),

    # ── GRAPE (4) ──
    "Grape___Black_rot"                                     : ("Grape",      "Black Rot",                 "Severe",    "Fungal (Guignardia bidwellii)",          "Prune infected clusters. Apply mancozeb or captan.",                    0.80),
    "Grape___Esca_(Black_Measles)"                          : ("Grape",      "Esca (Black Measles)",      "Severe",    "Fungal (Phaeoacremonium)",               "Remove infected wood. No effective chemical cure.",                     0.85),
    "Grape___Leaf_blight_(Isariopsis_Leaf_Spot)"            : ("Grape",      "Leaf Blight",               "Moderate",  "Fungal (Isariopsis clavispora)",         "Apply copper fungicide. Improve canopy airflow.",                       0.60),
    "Grape___healthy"                                       : ("Grape",      "Healthy",                   "Healthy ✅","None",                                  "No treatment needed.",                                                  0.00),

    # ── ORANGE (1) ──
    "Orange___Haunglongbing_(Citrus_greening)"              : ("Orange",     "Citrus Greening (HLB)",     "Critical",  "Bacterial (Candidatus Liberibacter)",    "No cure. Remove infected trees. Control psyllid insects.",              0.98),

    # ── PEACH (2) ──
    "Peach___Bacterial_spot"                                : ("Peach",      "Bacterial Spot",            "Moderate",  "Bacterial (Xanthomonas arboricola)",     "Apply copper bactericide. Avoid overhead watering.",                    0.65),
    "Peach___healthy"                                       : ("Peach",      "Healthy",                   "Healthy ✅","None",                                  "No treatment needed.",                                                  0.00),

    # ── PEPPER (2) ──
    "Pepper,_bell___Bacterial_spot"                         : ("Pepper",     "Bacterial Spot",            "Moderate",  "Bacterial (Xanthomonas campestris)",     "Copper bactericide spray. Use disease-free seeds.",                     0.65),
    "Pepper,_bell___healthy"                                : ("Pepper",     "Healthy",                   "Healthy ✅","None",                                  "No treatment needed.",                                                  0.00),

    # ── POTATO (3) ──
    "Potato___Early_blight"                                 : ("Potato",     "Early Blight",              "Mild",      "Fungal (Alternaria solani)",             "Apply azoxystrobin. Rotate crops annually.",                            0.50),
    "Potato___Late_blight"                                  : ("Potato",     "Late Blight",               "Critical",  "Oomycete (Phytophthora infestans)",      "Emergency fungicide application. Halt irrigation immediately.",         0.95),
    "Potato___healthy"                                      : ("Potato",     "Healthy",                   "Healthy ✅","None",                                  "No treatment needed.",                                                  0.00),

    # ── RASPBERRY (1) ──
    "Raspberry___healthy"                                   : ("Raspberry",  "Healthy",                   "Healthy ✅","None",                                  "No treatment needed.",                                                  0.00),

    # ── SOYBEAN (1) ──
    "Soybean___healthy"                                     : ("Soybean",    "Healthy",                   "Healthy ✅","None",                                  "No treatment needed.",                                                  0.00),

    # ── SQUASH (1) ──
    "Squash___Powdery_mildew"                               : ("Squash",     "Powdery Mildew",            "Moderate",  "Fungal (Podosphaera xanthii)",           "Apply potassium bicarbonate or neem oil spray.",                        0.55),

    # ── STRAWBERRY (2) ──
    "Strawberry___Leaf_scorch"                              : ("Strawberry", "Leaf Scorch",               "Moderate",  "Fungal (Diplocarpon earlianum)",         "Remove infected leaves. Apply captan fungicide.",                       0.60),
    "Strawberry___healthy"                                  : ("Strawberry", "Healthy",                   "Healthy ✅","None",                                  "No treatment needed.",                                                  0.00),

    # ── TOMATO (10) ──
    "Tomato___Bacterial_spot"                               : ("Tomato",     "Bacterial Spot",            "Moderate",  "Bacterial (Xanthomonas spp.)",           "Copper bactericide. Avoid overhead watering.",                          0.70),
    "Tomato___Early_blight"                                 : ("Tomato",     "Early Blight",              "Moderate",  "Fungal (Alternaria solani)",             "Remove lower leaves. Apply chlorothalonil or mancozeb.",                0.55),
    "Tomato___Late_blight"                                  : ("Tomato",     "Late Blight",               "Severe",    "Oomycete (Phytophthora infestans)",      "Immediate copper spray. Destroy infected plants.",                      0.90),
    "Tomato___Leaf_Mold"                                    : ("Tomato",     "Leaf Mold",                 "Moderate",  "Fungal (Passalora fulva)",               "Improve ventilation. Apply mancozeb.",                                  0.60),
    "Tomato___Septoria_leaf_spot"                           : ("Tomato",     "Septoria Leaf Spot",        "Moderate",  "Fungal (Septoria lycopersici)",          "Remove lower leaves. Apply chlorothalonil.",                            0.55),
    "Tomato___Spider_mites Two-spotted_spider_mite"         : ("Tomato",     "Spider Mites",              "Moderate",  "Pest (Tetranychus urticae)",             "Apply miticide or neem oil. Increase humidity.",                        0.60),
    "Tomato___Target_Spot"                                  : ("Tomato",     "Target Spot",               "Moderate",  "Fungal (Corynespora cassiicola)",        "Apply azoxystrobin or difenoconazole fungicide.",                       0.60),
    "Tomato___Tomato_Yellow_Leaf_Curl_Virus"                : ("Tomato",     "Yellow Leaf Curl Virus",    "Severe",    "Viral (Begomovirus)",                    "No cure. Remove infected plants. Control whiteflies.",                  0.88),
    "Tomato___Tomato_mosaic_virus"                          : ("Tomato",     "Mosaic Virus",              "Severe",    "Viral (Tobamovirus)",                    "Remove infected plants. Disinfect tools. No chemical cure.",            0.85),
    "Tomato___healthy"                                      : ("Tomato",     "Healthy",                   "Healthy ✅","None",                                  "No treatment needed.",                                                  0.00),
}

# Count check
healthy  = sum(1 for k in DISEASE_DB if "healthy" in k.lower())
diseased = len(DISEASE_DB) - healthy
print(f"✅ Disease DB loaded: {len(DISEASE_DB)} total classes")
print(f"   🟢 Healthy  : {healthy} classes")
print(f"   🟥 Diseased : {diseased} classes")

def get_info(class_name):
    if class_name in DISEASE_DB:
        p, d, stage, cause, remedy, sev = DISEASE_DB[class_name]
        return {"plant": p, "disease": d, "stage": stage,
                "cause": cause, "remedy": remedy, "severity": sev}
    # Fallback (should never trigger now)
    plant   = class_name.split("___")[0].replace("_", " ") if "___" in class_name else class_name
    disease = class_name.split("___")[1].replace("_", " ") if "___" in class_name else "Unknown"
    healthy = "healthy" in class_name.lower()
    return {"plant": plant,
            "disease": "Healthy" if healthy else disease,
            "stage":   "Healthy ✅" if healthy else "Detected",
            "cause":   "None" if healthy else "Pathogen",
            "remedy":  "No treatment needed." if healthy else "Consult agricultural expert.",
            "severity": 0.0 if healthy else 0.6}


# ──────────────────────────────
# CELL 5 — LOAD DATASET (Direct, No Download!)
# ──────────────────────────────
print("📥 Loading PlantVillage dataset via TensorFlow Datasets...")
print("   (This streams directly — no Kaggle account needed)\n")

(ds_train_raw, ds_valid_raw), info = tfds.load(
    'plant_village',
    split=['train[:80%]', 'train[80%:]'],
    as_supervised=True,
    with_info=True,
    shuffle_files=True
)

NUM_CLASSES  = info.features['label'].num_classes
CLASS_NAMES  = info.features['label'].names

# Save class index mapping
class_indices = {name: i for i, name in enumerate(CLASS_NAMES)}
with open(CONFIG["CLASS_JSON"], "w") as f:
    json.dump(class_indices, f, indent=2)

print(f"✅ Total disease classes : {NUM_CLASSES}")
print(f"✅ Total train images    : {info.splits['train'].num_examples}")
print(f"✅ Class indices saved   : {CONFIG['CLASS_JSON']}")
print(f"\n🌿 Sample classes:")
for name in CLASS_NAMES[:6]:
    print(f"   • {name}")
print(f"   ... and {NUM_CLASSES - 6} more")


# ───────────────────────────────────
# CELL 6 — DATA PIPELINE + AUGMENTATION
# ──────────────────────────────
IMG_SIZE   = CONFIG["IMG_SIZE"]
BATCH_SIZE = CONFIG["BATCH_SIZE"]

@tf.function
def preprocess_train(image, label):
    image = tf.image.resize(image, [IMG_SIZE, IMG_SIZE])
    image = tf.cast(image, tf.float32) / 255.0
    image = tf.image.random_flip_left_right(image)
    image = tf.image.random_flip_up_down(image)
    image = tf.image.random_brightness(image, max_delta=0.2)
    image = tf.image.random_contrast(image, lower=0.8, upper=1.2)
    image = tf.image.random_saturation(image, lower=0.8, upper=1.2)
    image = tf.image.random_hue(image, max_delta=0.1)
    image = tf.clip_by_value(image, 0.0, 1.0)
    label = tf.one_hot(label, NUM_CLASSES)
    return image, label

@tf.function
def preprocess_valid(image, label):
    image = tf.image.resize(image, [IMG_SIZE, IMG_SIZE])
    image = tf.cast(image, tf.float32) / 255.0
    label = tf.one_hot(label, NUM_CLASSES)
    return image, label

# Limit to ~5000 samples
ds_train = (ds_train_raw
            .take(CONFIG["TRAIN_SAMPLES"])
            .map(preprocess_train, num_parallel_calls=tf.data.AUTOTUNE)
            .shuffle(buffer_size=1000, seed=CONFIG["SEED"])
            .batch(BATCH_SIZE)
            .prefetch(tf.data.AUTOTUNE))

ds_valid = (ds_valid_raw
            .take(CONFIG["VALID_SAMPLES"])
            .map(preprocess_valid, num_parallel_calls=tf.data.AUTOTUNE)
            .batch(BATCH_SIZE)
            .prefetch(tf.data.AUTOTUNE))

print(f"✅ Train batches : {len(list(ds_train))}")
print(f"✅ Valid batches : {len(list(ds_valid))}")

# Preview augmented samples
plt.figure(figsize=(14, 4))
plt.suptitle("🌿 Sample Training Images (Augmented)", fontsize=13, fontweight='bold')
for images, labels in ds_train.take(1):
    for i in range(6):
        ax = plt.subplot(1, 6, i + 1)
        plt.imshow(images[i].numpy())
        lbl = CLASS_NAMES[tf.argmax(labels[i]).numpy()]
        plt.title(lbl.split("___")[-1][:15], fontsize=7)
        plt.axis("off")
plt.tight_layout()
plt.savefig("sample_images.png", dpi=120, bbox_inches='tight')
plt.show()


# ───────────────────────────────────
# CELL 7 — BUILD MODEL (MobileNetV2)
# ───────────────────────────────────
def build_model(num_classes, img_size=224):
    base = MobileNetV2(weights='imagenet', include_top=False,
                       input_shape=(img_size, img_size, 3))
    base.trainable = False  # Freeze initially

    inputs = tf.keras.Input(shape=(img_size, img_size, 3))
    x = base(inputs, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dense(512, activation='relu')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.4)(x)
    x = layers.Dense(256, activation='relu')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.3)(x)
    outputs = layers.Dense(num_classes, activation='softmax')(x)

    model = tf.keras.Model(inputs, outputs, name="PlantDiseaseDetector")
    model.compile(
        optimizer=optimizers.Adam(CONFIG["LEARNING_RATE"]),
        loss='categorical_crossentropy',
        metrics=['accuracy', tf.keras.metrics.TopKCategoricalAccuracy(k=3, name='top3_acc')]
    )

    trainable = sum([tf.size(w).numpy() for w in model.trainable_weights])
    total     = sum([tf.size(w).numpy() for w in model.weights])
    print(f"\n🧠 Model Architecture : MobileNetV2 + Custom Head")
    print(f"   Total parameters   : {total:,}")
    print(f"   Trainable params   : {trainable:,}")
    print(f"   Input shape        : ({img_size}, {img_size}, 3)")
    print(f"   Output classes     : {num_classes}")
    return model, base


model, base_model = build_model(NUM_CLASSES, IMG_SIZE)


# ──────────────────────────────
# CELL 8 — TRAIN MODEL (2 Phases)
# ──────────────────────────────

cb_list = [
    callbacks.EarlyStopping(monitor='val_accuracy', patience=5,
                            restore_best_weights=True, verbose=1),
    callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5,
                                patience=3, min_lr=1e-7, verbose=1),
    callbacks.ModelCheckpoint(CONFIG["MODEL_PATH"], save_best_only=True,
                              monitor='val_accuracy', verbose=1),
]

# ── PHASE 1: Train only the custom head ──
print("\n🚀 PHASE 1 — Training classification head (base frozen)...")
h1 = model.fit(ds_train, validation_data=ds_valid,
               epochs=CONFIG["EPOCHS_FROZEN"], callbacks=cb_list, verbose=1)

# ── PHASE 2: Fine-tune top layers of MobileNetV2 ──
print("\n🔧 PHASE 2 — Fine-tuning top MobileNetV2 layers...")
base_model.trainable = True
for layer in base_model.layers[:100]:    # keep bottom 100 layers frozen
    layer.trainable = False

model.compile(
    optimizer=optimizers.Adam(CONFIG["FINETUNE_LR"]),
    loss='categorical_crossentropy',
    metrics=['accuracy', tf.keras.metrics.TopKCategoricalAccuracy(k=3, name='top3_acc')]
)
h2 = model.fit(ds_train, validation_data=ds_valid,
               epochs=CONFIG["EPOCHS_FINETUNE"], callbacks=cb_list, verbose=1)

# Merge histories
history = {}
for k in h1.history:
    history[k] = h1.history[k] + h2.history.get(k, [])

print(f"\n✅ Best validation accuracy: {max(history['val_accuracy'])*100:.2f}%")


# ───────────────────────────────────
# CELL 9 — PLOT TRAINING CURVES
# ───────────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("🌿 Plant Disease Detection — Training Results", fontsize=15, fontweight='bold')

plots = [
    ('accuracy',  'val_accuracy',  'Accuracy',     '#2ecc71', '#27ae60'),
    ('loss',      'val_loss',      'Loss',          '#e74c3c', '#c0392b'),
    ('top3_acc',  'val_top3_acc',  'Top-3 Accuracy','#3498db', '#2980b9'),
]

for ax, (tk, vk, title, c1, c2) in zip(axes, plots):
    phase1_end = len(h1.history[tk])
    epochs_all = range(1, len(history[tk]) + 1)

    ax.axvspan(1, phase1_end, alpha=0.08, color='yellow', label='Phase 1')
    ax.axvspan(phase1_end + 1, len(history[tk]), alpha=0.08, color='cyan', label='Phase 2')
    ax.plot(epochs_all, history[tk],  color=c1, lw=2.5, label='Train')
    ax.plot(epochs_all, history[vk],  color=c2, lw=2.5, ls='--', label='Validation')
    ax.set_title(title, fontsize=13, fontweight='bold')
    ax.set_xlabel('Epoch')
    ax.set_ylabel(title)
    ax.legend()
    ax.grid(True, alpha=0.3)
    ax.set_facecolor('#f8f9fa')

plt.tight_layout()
plt.savefig("training_curves.png", dpi=150, bbox_inches='tight')
plt.show()
print("✅ Saved: training_curves.png")


# ──────────────────────────────
# CELL 10 — EVALUATE + CONFUSION MATRIX
# ──────────────────────────────
print("📊 Evaluating model on validation set...")
loss, acc, top3 = model.evaluate(ds_valid, verbose=1)
print(f"\n✅ Validation Accuracy  : {acc*100:.2f}%")
print(f"✅ Top-3 Accuracy       : {top3*100:.2f}%")
print(f"✅ Validation Loss      : {loss:.4f}")

# Get predictions for confusion matrix
y_true_list, y_pred_list = [], []
for images, labels in ds_valid:
    preds  = model.predict(images, verbose=0)
    y_pred_list.extend(np.argmax(preds, axis=1))
    y_true_list.extend(np.argmax(labels.numpy(), axis=1))

y_true = np.array(y_true_list)
y_pred = np.array(y_pred_list)

# Plot confusion matrix (top 15 classes for readability)
top_n = 15
counts = np.bincount(y_true)
top_idx = np.argsort(counts)[-top_n:]
mask = np.isin(y_true, top_idx)
cm = confusion_matrix(y_true[mask], y_pred[mask], labels=top_idx)
cm_norm = cm.astype('float') / (cm.sum(axis=1, keepdims=True) + 1e-8)
short = [CLASS_NAMES[i].split("___")[-1][:18] for i in top_idx]

fig, ax = plt.subplots(figsize=(14, 12))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlOrRd',
            xticklabels=short, yticklabels=short, ax=ax,
            linewidths=0.5, cbar_kws={'label': 'Normalized Frequency'})
ax.set_title("🌿 Disease Classification — Confusion Matrix (Top 15 Classes)",
             fontsize=13, fontweight='bold')
ax.set_xlabel("Predicted", fontsize=11)
ax.set_ylabel("True Label", fontsize=11)
plt.xticks(rotation=45, ha='right', fontsize=9)
plt.yticks(rotation=0, fontsize=9)
plt.tight_layout()
plt.savefig("confusion_matrix.png", dpi=150, bbox_inches='tight')
plt.show()

# Classification report
print("\n📋 Classification Report:\n")
print(classification_report(y_true, y_pred,
      target_names=CLASS_NAMES, zero_division=0))


# ───────────────────────────────────
# CELL 11 — GRAD-CAM (Visual Explainability)
# ──────────────────────────────
def get_gradcam(model, img_array, layer_name="Conv_1"):
    """Highlight which part of the leaf the AI focused on."""
    grad_model = tf.keras.Model(
        inputs=model.inputs,
        outputs=[model.get_layer(layer_name).output, model.output]
    )
    with tf.GradientTape() as tape:
        conv_out, preds = grad_model(img_array)
        top_class = tf.argmax(preds[0])
        score = preds[:, top_class]
    grads = tape.gradient(score, conv_out)
    pooled = tf.reduce_mean(grads, axis=(0, 1, 2))
    heatmap = conv_out[0] @ pooled[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)
    heatmap = tf.maximum(heatmap, 0)
    heatmap = heatmap / (tf.math.reduce_max(heatmap) + 1e-8)
    return heatmap.numpy()

def overlay_cam(img_np, heatmap, alpha=0.45):
    h = cv2.resize(heatmap, (img_np.shape[1], img_np.shape[0]))
    h_color = cv2.applyColorMap(np.uint8(255 * h), cv2.COLORMAP_JET)
    h_color = cv2.cvtColor(h_color, cv2.COLOR_BGR2RGB)
    return cv2.addWeighted(img_np, 1 - alpha, h_color, alpha, 0)


# ──────────────────────────────
# CELL 12 — PREDICT ON RANDOM VALIDATION IMAGE
# ──────────────────────────────
def predict_and_show(model, dataset, class_names, num_examples=3):
    """Run prediction on random images and display results with Grad-CAM."""

    fig, axes = plt.subplots(num_examples, 3, figsize=(15, 5 * num_examples))
    fig.patch.set_facecolor('#0d1117')
    fig.suptitle("🛿 UAV Plant Disease Detection Results",
                 color='white', fontsize=16, fontweight='bold', y=1.01)

    idx_to_class = {v: k for k, v in enumerate(class_names)}

    sample_images, sample_labels = [], []
    for imgs, lbls in dataset.take(1):
        for i in range(min(num_examples, len(imgs))):
            sample_images.append(imgs[i].numpy())
            sample_labels.append(np.argmax(lbls[i].numpy()))

    for row, (img, true_idx) in enumerate(zip(sample_images, sample_labels)):
        img_batch = np.expand_dims(img, axis=0)
        preds     = model.predict(img_batch, verbose=0)[0]
        top3_idx  = np.argsort(preds)[::-1][:3]
        pred_idx  = top3_idx[0]

        true_class = class_names[true_idx]
        pred_class = class_names[pred_idx]
        info       = get_info(pred_class)
        conf       = preds[pred_idx]
        correct    = (true_idx == pred_idx)

        # ── Col 1: Original image ──
        ax1 = axes[row][0]
        ax1.imshow(img)
        ax1.set_facecolor('#161b22')
        ax1.set_title(f"📸 True: {true_class.split('___')[-1]}",
                      color='white', fontsize=9, pad=6)
        ax1.axis('off')

        # ── Col 2: Grad-CAM overlay ──
        ax2 = axes[row][1]
        try:
            heatmap  = get_gradcam(model, img_batch)
            img_uint = (img * 255).astype(np.uint8)
            overlaid = overlay_cam(img_uint, heatmap)
            ax2.imshow(overlaid)
            ax2.set_title("🔥 Grad-CAM: Disease Region",
                          color='white', fontsize=9, pad=6)
        except Exception:
            ax2.imshow(img)
            ax2.set_title("📷 Processed Image", color='white', fontsize=9, pad=6)
        ax2.axis('off')

        # ── Col 3: Prediction panel ──
        ax3 = axes[row][2]
        ax3.set_facecolor('#161b22')
        ax3.axis('off')

        is_healthy  = info['disease'] == 'Healthy'
        status_col  = '#2ecc71' if is_healthy else '#e74c3c'
        result_icon = '✅ CORRECT' if correct else '❌ WRONG'
        result_col  = '#2ecc71' if correct else '#e74c3c'

        lines = [
            (f"{'✅' if is_healthy else '🚨'} {info['disease']}",  status_col,  14, 'bold'),
            (f"🌱 Plant    : {info['plant']}",                      '#58a6ff',   10, 'normal'),
            (f"📊 Confidence: {conf*100:.1f}%",                     '#f0f6fc',   10, 'normal'),
            (f"⚠️  Stage    : {info['stage']}",                     '#ffa657',   10, 'normal'),
            (f"🦠 Cause    : {info['cause']}",                      '#c9d1d9',   9,  'normal'),
            ("",                                                     'white',     6,  'normal'),
            ("💊 Recommended Action:",                              '#ffa657',   10, 'bold'),
            (f"  {info['remedy']}",                                 '#c9d1d9',   9,  'normal'),
            ("",                                                     'white',     6,  'normal'),
            (f"Result: {result_icon}",                              result_col,  11, 'bold'),
        ]

        y = 0.97
        for text, color, size, weight in lines:
            ax3.text(0.03, y, text, transform=ax3.transAxes, color=color,
                     fontsize=size, fontweight=weight, va='top',
                     wrap=True, clip_on=True)
            y -= 0.095

        # Severity bar
        if not is_healthy and y > 0.05:
            sev = info['severity']
            bar_col = '#2ecc71' if sev < 0.4 else '#f39c12' if sev < 0.7 else '#e74c3c'
            ax3.text(0.03, y, f"Severity: {sev*100:.0f}%", transform=ax3.transAxes,
                     color='#ffa657', fontsize=9, fontweight='bold', va='top')
            y -= 0.07
            ax3.add_patch(mpatches.FancyBboxPatch(
                (0.03, y), 0.90, 0.04, boxstyle="round,pad=0.01",
                facecolor='#21262d', transform=ax3.transAxes, zorder=1))
            ax3.add_patch(mpatches.FancyBboxPatch(
                (0.03, y), sev * 0.90, 0.04, boxstyle="round,pad=0.01",
                facecolor=bar_col, transform=ax3.transAxes, zorder=2))

        # Top-3 bar
        y -= 0.09
        ax3.text(0.03, y, "📈 Top-3 Predictions:", transform=ax3.transAxes,
                 color='#8b949e', fontsize=9, fontweight='bold', va='top')
        for rank, idx in enumerate(top3_idx):
            y -= 0.075
            marker = "▶" if rank == 0 else "  "
            ax3.text(0.03, y,
                     f"{marker} {class_names[idx].split('___')[-1][:22]:<24} {preds[idx]*100:5.1f}%",
                     transform=ax3.transAxes,
                     color='#f0f6fc' if rank == 0 else '#8b949e',
                     fontsize=8,
                     fontweight='bold' if rank == 0 else 'normal',
                     va='top', fontfamily='monospace')

    plt.tight_layout()
    plt.savefig("prediction_results.png", dpi=150, bbox_inches='tight',
                facecolor='#0d1117', edgecolor='none')
    plt.show()
    print("✅ Saved: prediction_results.png")


# Run predictions on 3 random validation images
predict_and_show(model, ds_valid, CLASS_NAMES, num_examples=3)


# ──────────────────────────────
# CELL 13 — EXPORT TO TFLite (RASPBERRY PI)
# ──────────────────────────────
print("\n📦 Exporting model to TFLite for Raspberry Pi...")

converter = tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
converter.target_spec.supported_types = [tf.float16]
tflite_model = converter.convert()

with open(CONFIG["TFLITE_PATH"], 'wb') as f:
    f.write(tflite_model)

orig_mb = os.path.getsize(CONFIG["MODEL_PATH"]) / 1e6
lite_mb  = os.path.getsize(CONFIG["TFLITE_PATH"]) / 1e6
print(f"✅ Keras model  : {CONFIG['MODEL_PATH']}  ({orig_mb:.1f} MB)")
print(f"✅ TFLite model : {CONFIG['TFLITE_PATH']} ({lite_mb:.1f} MB)")
print(f"✅ Size reduced : {(1 - lite_mb/orig_mb)*100:.0f}%  ← smaller = faster on Pi")


# ──────────────────────────────
# CELL 14 — SAVE RASPBERRY PI SCRIPT
# ──────────────────────────────
PI_SCRIPT = '''#!/usr/bin/env python3
"""
🛿 Raspberry Pi Live Plant Disease Detector
Runs on Pi 4 with Pi Camera Module or USB webcam

Install:
    pip3 install tflite-runtime opencv-python numpy

Run:
    python3 raspberry_pi_detector.py
"""
import numpy as np
import json
import cv2
import tflite_runtime.interpreter as tflite

MODEL_PATH  = "plant_disease_model.tflite"
CLASS_JSON  = "class_indices.json"
IMG_SIZE    = 224
MIN_CONF    = 0.50   # Minimum confidence threshold

# ── Load model ──
interpreter = tflite.Interpreter(model_path=MODEL_PATH, num_threads=4)
interpreter.allocate_tensors()
inp  = interpreter.get_input_details()
outp = interpreter.get_output_details()

# ── Load class labels ──
with open(CLASS_JSON) as f:
    class_indices = json.load(f)
idx_to_class = {v: k for k, v in class_indices.items()}

def predict_frame(frame):
    img = cv2.resize(frame, (IMG_SIZE, IMG_SIZE))
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB).astype(np.float32) / 255.0
    interpreter.set_tensor(inp[0]["index"], np.expand_dims(img, 0))
    interpreter.invoke()
    return interpreter.get_tensor(outp[0]["index"])[0]

def get_color(class_name, conf):
    if conf < MIN_CONF:     return (0, 165, 255)   # orange = low confidence
    if "healthy" in class_name.lower(): return (0, 255, 0)  # green = healthy
    return (0, 0, 255)                              # red = disease

# ── Start camera ──
cap = cv2.VideoCapture(0)   # 0 = default camera
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 640)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 480)

print("🛿 UAV Plant Disease Detector — Press Q to quit")

frame_count = 0
result_text = "Initializing..."
result_color = (255, 255, 255)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Run inference every 5 frames (saves CPU on Pi)
    if frame_count % 5 == 0:
        preds     = predict_frame(frame)
        top_idx   = int(np.argmax(preds))
        conf      = float(preds[top_idx])
        cls_name  = idx_to_class.get(top_idx, "Unknown")
        disease   = cls_name.split("___")[-1].replace("_", " ") if "___" in cls_name else cls_name
        plant     = cls_name.split("___")[0].replace("_", " ") if "___" in cls_name else ""

        if conf < MIN_CONF:
            result_text  = f"Low confidence ({conf*100:.0f}%) — move closer"
        else:
            result_text  = f"{plant}: {disease}  ({conf*100:.1f}%)"
        result_color = get_color(cls_name, conf)

    frame_count += 1

    # ── Draw overlay ──
    cv2.rectangle(frame, (0, 0), (frame.shape[1], 65), (0, 0, 0), -1)
    cv2.putText(frame, "UAV Disease Detector", (10, 22),
                cv2.FONT_HERSHEY_SIMPLEX, 0.65, (200, 200, 200), 1)
    cv2.putText(frame, result_text, (10, 52),
                cv2.FONT_HERSHEY_SIMPLEX, 0.75, result_color, 2)

    cv2.imshow("Plant Disease Detection", frame)
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
'''

with open("raspberry_pi_detector.py", "w") as f:
    f.write(PI_SCRIPT)
print("✅ Saved: raspberry_pi_detector.py")


# ──────────────────────────────
# CELL 15 — DOWNLOAD ALL FILES TO YOUR PC
# ──────────────────────────────
from google.colab import files

print("\n📥 Downloading all output files...")
for fname in [
    CONFIG["MODEL_PATH"],
    CONFIG["TFLITE_PATH"],
    CONFIG["CLASS_JSON"],
    "raspberry_pi_detector.py",
    "training_curves.png",
    "confusion_matrix.png",
    "prediction_results.png",
    "sample_images.png",
]:
    if os.path.exists(fname):
        files.download(fname)
        print(f"   ⬇️  {fname}")
    else:
        print(f"   ⚠️  {fname} not found (may not have been generated)")

print("\n" + "="*60)
print("  🎉 COMPLETE! Files downloaded to your PC.")
print("="*60)
print(f"  📦 Keras model       : {CONFIG['MODEL_PATH']}")
print(f"  📦 TFLite (Pi ready) : {CONFIG['TFLITE_PATH']}")
print(f"  📦 Class labels      : {CONFIG['CLASS_JSON']}")
print(f"  📦 Pi camera script  : raspberry_pi_detector.py")
print(f"  📊 Best val accuracy : {max(history['val_accuracy'])*100:.2f}%")
print("="*60)
print("\n🛿 Copy .tflite + class_indices.json + raspberry_pi_detector.py")
print("   to your Raspberry Pi and run:")
print("   $ python3 raspberry_pi_detector.py")

# ──────────────────────────────
# CELL 16 — UPLOAD YOUR OWN IMAGE & PREDICT
# ──────────────────────────────
from google.colab import files
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from PIL import Image
import io

# Step 1: Upload your image
print("📁 Click 'Choose Files' and upload any leaf image...")
uploaded = files.upload()

# Step 2: Predict on it
for filename, data in uploaded.items():
    print(f"\n🔍 Analyzing: {filename}")

    # Load image
    img = Image.open(io.BytesIO(data)).convert("RGB")
    img_resized = img.resize((224, 224))
    img_array = np.array(img_resized).astype(np.float32) / 255.0
    img_batch = np.expand_dims(img_array, axis=0)

    # Predict
    preds    = model.predict(img_batch, verbose=0)[0]
    top3_idx = np.argsort(preds)[::-1][:3]
    pred_idx = top3_idx[0]
    conf     = preds[pred_idx]
    cls_name = CLASS_NAMES[pred_idx]
    info     = get_info(cls_name)

    # Show result
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    fig.patch.set_facecolor('#0d1117')

    # Left: uploaded image
    axes[0].imshow(img)
    axes[0].set_title("📸 Your Uploaded Leaf", color='white', fontsize=12)
    axes[0].axis('off')

    # Right: prediction info
    axes[1].set_facecolor('#161b22')
    axes[1].axis('off')

    is_healthy = info['disease'] == 'Healthy'
    color = '#2ecc71' if is_healthy else '#e74c3c'
    icon  = '✅' if is_healthy else '🚨'

    lines = [
        (f"{icon}  {info['disease']}",           color,     16, 'bold'),
        (f"🌱 Plant      : {info['plant']}",      '#58a6ff', 11, 'normal'),
        (f"📊 Confidence : {conf*100:.1f}%",      '#f0f6fc', 11, 'normal'),
        (f"⚠️  Stage      : {info['stage']}",     '#ffa657', 11, 'normal'),
        (f"🦠 Cause      : {info['cause']}",      '#c9d1d9', 10, 'normal'),
        ("",                                       'white',   6,  'normal'),
        ("💊 Recommended Action:",                '#ffa657', 11, 'bold'),
        (f"   {info['remedy']}",                  '#c9d1d9', 10, 'normal'),
        ("",                                       'white',   6,  'normal'),
        ("📈 Top-3 Predictions:",                 '#8b949e', 10, 'bold'),
    ]

    y = 0.97
    for text, col, size, weight in lines:
        axes[1].text(0.05, y, text, transform=axes[1].transAxes,
                     color=col, fontsize=size, fontweight=weight, va='top')
        y -= 0.09

    for rank, idx in enumerate(top3_idx):
        marker = "▶" if rank == 0 else "  "
        axes[1].text(0.05, y,
                     f"{marker} {CLASS_NAMES[idx].split('___')[-1][:24]:<26} {preds[idx]*100:5.1f}%",
                     transform=axes[1].transAxes,
                     color='#f0f6fc' if rank == 0 else '#8b949e',
                     fontsize=9, fontweight='bold' if rank == 0 else 'normal',
                     va='top', fontfamily='monospace')
        y -= 0.08

    plt.suptitle("🛿 UAV Plant Disease Detection", color='white',
                 fontsize=14, fontweight='bold')
    plt.tight_layout()
    plt.savefig("my_prediction.png", dpi=150, bbox_inches='tight',
                facecolor='#0d1117')
    plt.show()

    print("\n" + "="*50)
    print(f"  Disease    : {info['disease']}")
    print(f"  Plant      : {info['plant']}")
    print(f"  Confidence : {conf*100:.1f}%")
    print(f"  Stage      : {info['stage']}")
    print(f"  Remedy     : {info['remedy']}")
    print("="*50)
```
